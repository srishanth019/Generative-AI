```python
import requests
import re
```

```python
def check_prompt(prompt):
    prompt = prompt.lower().strip()

    injection_patterns = [
        r"ignore.*instructions",
        r"ignore.*guidelines",
        r"forget.*instructions",
        r"override.*instructions",
        r"reveal.*system prompt",
        r"show.*system prompt",
        r"display.*system prompt",
        r"reveal.*instructions",
        r"bypass.*restrictions",
        r"bypass.*safety",
        r"act as.*without restrictions",
        r"jailbreak"
    ]

    unsafe_patterns = [
        r"how.*kill.*person",
        r"how.*kill.*someone",
        r"how.*murder.*person",
        r"how.*murder.*someone",
        r"how.*harm.*person",
        r"how.*harm.*someone",

        r"how.*make.*bomb",
        r"how.*build.*bomb",
        r"create.*malware",
        r"write.*ransomware",

        r"hack.*password",
        r"steal.*password",

        r"kill.*person",
        r"kill.*someone",
        r"harm.*person",
        r"harm.*someone"
    ]

    for pattern in injection_patterns:
        if re.search(pattern, prompt):
            return False, "Potential Prompt Injection Detected: Attempt to override system instructions."

    for pattern in unsafe_patterns:
        if re.search(pattern, prompt):
            return False, "Unsafe Instruction Detected: The prompt requests potentially harmful information."

    return True, "Prompt is safe."
```

```python
def ask_ollama(prompt):
    url = "http://localhost:11434/api/generate"

    data = {
        "model": "phi3:mini",
        "prompt": prompt,
        "stream": False
    }

    try:
        response = requests.post(url, json=data)

        if response.status_code == 200:
            result = response.json()
            return result["response"]
        else:
            return "Error communicating with Ollama."

    except requests.exceptions.ConnectionError:
        return "Could not connect to Ollama. Make sure Ollama is running."
```

```python
user_prompt = input("Enter your prompt: ")

is_safe, reason = check_prompt(user_prompt)

print("\n--- Guardrail Result ---")

if is_safe:
    print("Safe")
    print("Reason:", reason)

    print("\nSending prompt to Ollama...")

    response = ask_ollama(user_prompt)

    print("\n--- LLM Response ---")
    print(response)

else:
    print("Unsafe")
    print("Reason:", reason)

    print("\nRequest not sent to the LLM.")
    print("\nRequest Blocked. This prompt violates AI safety guidelines.")
```
